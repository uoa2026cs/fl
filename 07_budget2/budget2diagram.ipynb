```python
import pandas
from matplotlib import pyplot
def flwr_accuracy(ax, file, name, start=0, key="accuracy", loc=None):
    df = pandas.read_csv(file + ".csv")
    for id_val, g in df.groupby("id", sort=False):
        g = g.sort_values("round").tail(len(g) - start)
        ax.plot(g["round"], g[key], label=str(id_val))

    ax.set_xlabel("round")
    ax.set_ylabel(name)
    ax.grid(True, alpha=0.3)
    if loc is not None:
        ax.legend(loc=loc)
fig, (ax1, ax2) = pyplot.subplots(1, 2, figsize=(13, 4))
flwr_accuracy(ax1, "budget2", "overall accuracy", 160)
flwr_accuracy(ax2, "budget2_sub", "disaster accuracy", 160, loc="upper left")
# flwr_accuracy(ax3, "budget2", 160, "loss")
fig.savefig("budget2diagram.pdf", bbox_inches="tight")
```
