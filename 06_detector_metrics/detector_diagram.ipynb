```python
from matplotlib import pyplot
import pandas
def flwr_values(ax, file, name, start=1, loc=None):
    df = pandas.read_csv(file + ".csv")
    df["id"] = df["id"].astype("string").str[:14]
    for id_val, g in df.groupby("id", sort=False):
        g = g.sort_values("round").tail(len(g) - start)
        ax.plot(g["round"], g["value"], label=str(id_val))
    ax.set_xlabel(name + ", round")
    ax.set_ylabel("value")
    ax.grid(True, alpha=0.3)
    if loc is not None:
        ax.legend(loc=loc)
fig, (ax1, ax2) = pyplot.subplots(1, 2, figsize=(13, 4))
flwr_values(ax1, "clear_norm", "norm surge", loc="upper left")
flwr_values(ax2, "clear_cos", "cosine divergence")
fig.savefig("detector_clear.pdf", bbox_inches="tight")
```
